```python
#### AGE DURATION CALCULATOR

def survival_duration(age):
    months = age * 12
    weeks = age * 52
    days = age * 365
    hours = days * 24
    minutes = hours * 60
    seconds = minutes * 60

    #### ABOVE ARE THE FORMULAS TO CALCULATE AGE IN DIFFERENT UNITS.
    #### DITIONARY IS DEFINED TO CALCULATE.
    
    return {
        'Months': months,
        'Weeks': weeks,
        'Days': days,
        'Hours': hours,
        'Minutes': minutes,
        'Seconds': seconds
    }

   
def calculate():
    try:

     ### TAKING INPUTS.
        ### Thought if user input is wrong then what to do thats why using here a if loop
        
        age = int(input("Enter Your Age: "))
        if age < 0:
            print("Error: Age cannot be negative.")
            return
        
        durations = survival_duration(age)
        
        print("\nThe person has lived:")
        for unit, value in durations.items():
            print(f"{value:,} {unit}") 

    except ValueError:
        print("Error: Please enter valid age.")

    
if __name__ == "__main__":
    
    calculate()
```
