### Write a regular expression to search for "python" case insensitive in the string line = "I Love Python"

```python
str = "I Love Python"
s = re.search("python",str,re.IGNORECASE)
print(s)
```

```python
### Write a regex to extract all the digits from the string line = "My Number is 12345"
```

```python
### Write a regex to find all words starting with a capital letter in the line = "Hello World zthis is Python"
```

```python
regex="Hello World This is Python"
pattern=r'\b[A-Z][a-z]*\b'
reg=re.findall(pattern,regex)
print(reg)
```

```python
### Write a regular Expression to check if the string line="Python is awesome" ends with awesome
```

```python
file="Python is awesome"
f=re.search(r"\b\w+\b$",file)
print(f)
```

```python
### Write a regular expression to find all occurrences of either "cat" or "dog" in the string line="The cat and the dog are friends"
```

```python
line="The cat and the dog are friends"
x=re.findall("cat|dog",line)
print(x)
```

```python
### Write a regex to remove all characters from the string line="Python123! Is fun" that are not alphabet letters (keep[A-Z] and a-z
```

```python
txt="Python123! Is Fun"
t=re.sub(r"[^A-Za-z]","",txt)
print(t)
```
